```python
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# 1. Calculate funnel stages
# -----------------------------

watched = df["videos_watched_pct"] > 50

assignment_ratio = (
    df["assignments_submitted_clean"] /
    df["assignments_total"]
)

most_assignments = assignment_ratio >= 0.80

completed = df["completed_flag"] == 1
opted_in = df["placement_assist_optin"] == 1
placed = df["placed_flag"] == 1

stages = [
    "Enrolled",
    "Watched >50%",
    "Submitted ≥80%\nAssignments",
    "Completed",
    "Placement Opt-in",
    "Placed"
]

values = [
    len(df),
    watched.sum(),
    (watched & most_assignments).sum(),
    (watched & most_assignments & completed).sum(),
    (watched & most_assignments & completed & opted_in).sum(),
    (watched & most_assignments & completed & opted_in & placed).sum()
]

# -----------------------------
# 2. Calculate percentages
# -----------------------------

percentages = [
    (value / values[0]) * 100
    for value in values
]

# Drop from previous stage
drops = ["—"]

for i in range(1, len(values)):
    drops.append(values[i-1] - values[i])

# -----------------------------
# 3. Create funnel chart
# -----------------------------

fig, ax = plt.subplots(figsize=(12, 8))

y = np.arange(len(stages))

# Normalize widths
max_value = max(values)
widths = np.array(values) / max_value

# Draw funnel bars
ax.barh(
    y,
    widths,
    height=0.75
)

# Put first stage at top
ax.invert_yaxis()

# -----------------------------
# 4. Add labels
# -----------------------------

for i in range(len(stages)):

    ax.text(
        widths[i] / 2,
        i,
        f"{values[i]:,}",
        ha="center",
        va="center",
        fontsize=14,
        fontweight="bold"
    )

    ax.text(
        1.02,
        i,
        f"{percentages[i]:.2f}%",
        va="center",
        fontsize=12
    )

    if i > 0:
        ax.text(
            1.15,
            i,
            f"Drop: {drops[i]:,}",
            va="center",
            fontsize=10
        )

# -----------------------------
# 5. Titles
# -----------------------------

ax.set_title(
    "SkillForge Learner Funnel",
    fontsize=22,
    fontweight="bold",
    pad=20
)

ax.text(
    0.5,
    1.02,
    "From Enrolment to Placement",
    transform=ax.transAxes,
    ha="center",
    fontsize=13
)

ax.set_yticks(y)
ax.set_yticklabels(stages, fontsize=12)

ax.set_xlabel(
    "Relative Funnel Size",
    fontsize=12
)

# Remove unnecessary borders
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_visible(False)

ax.set_xlim(0, 1.45)

plt.tight_layout()
plt.show()
```

```python
# Clean delivery mode
df["delivery_clean"] = (
    df["delivery_mode"]
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
)

delivery_map = {
    "self paced": "Self Paced",
    "self-paced": "Self Paced",
    "selfpace": "Self Paced",
    "live cohort": "Live Cohort",
    "live-cohort": "Live Cohort",
    "live": "Live Cohort",
    "hybrid": "Hybrid",
    "hybrid mode": "Hybrid"
}

df["delivery_clean"] = df["delivery_clean"].replace(delivery_map)
```
